# 01: Data Exploration & Feature Analysis

This notebook demonstrates flow distribution, feature dimensionality, and the statistical separation between normal and attack traffic.

```python
import sys
sys.path.insert(0, '../src')
import pickle
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from config import DATA_CONFIG
```

### Load Preprocessed Traffic Records

```python
with open(f'../{DATA_CONFIG["benign_train_path"]}', 'rb') as f:
    benign_records = pickle.load(f)
with open(f'../{DATA_CONFIG["attack_test_path"]}', 'rb') as f:
    attack_records = pickle.load(f)
print(f'Benign Flows: {len(benign_records):,}')
print(f'Attack Flows: {len(attack_records):,}')
print(f'Feature dimension per flow: {benign_records[0]["edge_features"].shape[0]}')
```

### Inspect Attack Categories

```python
attack_types = [r['attack_type'] for r in attack_records]
pd.Series(attack_types).value_counts().plot(kind='bar', color='darkred', figsize=(8, 4), title='Attack Class Distribution')
plt.ylabel('Flow Count')
plt.show()
```
